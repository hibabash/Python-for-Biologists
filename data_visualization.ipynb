Boot Strapping:mean of mean of random samples:

```python
list1=[]
for i in range(50): #50 means will be appended to the list
    sample1=iris.sample(30) #samplying only 30 rows
    mean1=round(sample1["Sepal.Length"].mean(),3)
    list1.append(mean1)

print(list1)
mean_of_means=sum(list1)/50
print(min(list1),max(list1))
print("Grand mean :",round(mean_of_means,2))
```

Examples: # Plot the mean of sepal length, species wise and standared error

```python
pwd
```

```python
iris=pd.read_csv(r"C:\Users\HP\iris_dataset.txt",sep=" ")
iris
```

```python
sl=iris[["Sepal.Length","Species"]] #selecting only the required columns that is subsetting
sl
```

```python
sl.head(5)
```

```python
#Calculating the mean and std in a single step
mean_std=sl.groupby(["Species"],as_index=False).agg({"Sepal.Length":['mean','std']}) #groupby: to group by species,agg:when calculating more than one parameter.
mean_std
```

```python
mean_std.columns=["Species","mean","std"] #renaming the columns
mean_std
```

```python
graph1=ggplot(mean_std)\
+ aes(x="Species",y="mean",fill="Species")\
+ geom_col()
graph1
```

```python
graph2=graph1+geom_errorbar(aes(x="Species",ymin=mean_std["mean"]-mean_std["std"]\
                               ,ymax=mean_std["mean"]+mean_std["std"], width = 0.05))
graph2
```

```python
graph2.save("exampleplot.png",dpi=1200)
```
