## **DNA composition and GC content**

```python
dna=input("Enter the DNA sequence :").upper()
length1=len(dna)
print(dna)

a=dna.count("A")
g=dna.count("G")
c=dna.count("C")
t=dna.count("T")

print(" A,G,C,T\n",a,g,c,t)
gc=((g+c)/length1)*100
print("GC content is ",gc)
```

```python
dna="AATTGGCCNN"
purine=0;pyrimidine=0;un=0
length1=len(dna)

for i in range(0,length1):
  nuc=dna[i]

  if nuc=="A" or nuc=="G":
    print(nuc,i,"purines")
    purine=purine+1

  elif nuc=="T" or nuc=="C":
    print(nuc,i,"pyrimidnes")
    pyrimidine+=1

  else:
    print(nuc,i,"un")
    un+=1

print("purines :",purine)
print("pyrimidines :",pyrimidine)
print("unknown :",un)
```

# **Annealing Temperature of a primer sequence**

```python
seq = input("Enter the primer sequence :").upper()
print(seq)
length1=len(seq)
print(length1)

nuc="ATGCN"

for i in nuc:
   if i=="G" or i=="C":
    z=(seq.count("G")+seq.count("C"))
   elif i=="A"or i=="T":
    x=(seq.count("A")+seq.count("T"))
   else:
    n=(seq.count("N"))
print(z)
print(x)
temp=4*z+2*x
print("Annealing temperature is :",temp)
```
